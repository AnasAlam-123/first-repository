```
import numpy as np
import matplotlib.pyplot as plt

def barnsley_fern(n_points=100000):
    x, y = 0, 0
    points = np.zeros((n_points, 2))

    for i in range(n_points):
        r = np.random.random()

        if r < 0.01:
            # Stem
            x_new, y_new = 0, 0.16 * y
        elif r < 0.86:
            # Successively smaller leaflets
            x_new = 0.85 * x + 0.04 * y
            y_new = -0.04 * x + 0.85 * y + 1.6
        elif r < 0.93:
            # Largest left-hand leaflet
            x_new = 0.20 * x - 0.26 * y
            y_new = 0.23 * x + 0.22 * y + 1.6
        else:
            # Largest right-hand leaflet
            x_new = -0.15 * x + 0.28 * y
            y_new = 0.26 * x + 0.24 * y + 0.44

        x, y = x_new, y_new
        points[i] = [x, y]

    return points

points = barnsley_fern(100000)

plt.figure(figsize=(6, 10))
plt.scatter(points[:, 0], points[:, 1], s=0.2, color="darkgreen")
plt.axis("equal")
plt.axis("off")
plt.title("Barnsley Fern")
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt

def koch_segment(p1, p2, depth):
    if depth == 0:
        return [p1, p2]

    p1 = np.array(p1, dtype=float)
    p2 = np.array(p2, dtype=float)

    delta = (p2 - p1) / 3.0
    a = p1
    b = p1 + delta
    d = p1 + 2 * delta
    e = p2

    angle = -np.pi / 3
    rot = np.array([
        [np.cos(angle), -np.sin(angle)],
        [np.sin(angle),  np.cos(angle)]
    ])
    c = b + rot.dot(d - b)

    points = []
    points += koch_segment(a, b, depth - 1)[:-1]
    points += koch_segment(b, c, depth - 1)[:-1]
    points += koch_segment(c, d, depth - 1)[:-1]
    points += koch_segment(d, e, depth - 1)

    return points

def koch_snowflake(depth, size=1.0):
    height = size * np.sqrt(3) / 2
    p1 = (0, 0)
    p2 = (size, 0)
    p3 = (size / 2, height)

    points = []
    points += koch_segment(p1, p2, depth)[:-1]
    points += koch_segment(p2, p3, depth)[:-1]
    points += koch_segment(p3, p1, depth)

    points = np.array(points)
    return points[:, 0], points[:, 1]

depth = 4
x, y = koch_snowflake(depth)

plt.figure(figsize=(6, 6))
plt.plot(x, y, color="teal")
plt.fill(x, y, color="lightblue", alpha=0.4)
plt.axis("equal")
plt.axis("off")
plt.title(f"Koch Snowflake (depth={depth})")
plt.show()
```

```
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.colors import LinearSegmentedColormap

# ----------------------------------------------------------------------------
# 1. FOSSIL COLOR PALETTES
# ----------------------------------------------------------------------------
# Warm amber/rust gradient used for the ammonite shells (dark core -> bright
# fossilized-resin highlight -> rust edge).
AMBER_CMAP = LinearSegmentedColormap.from_list(
    "fossil_amber",
    ["#160d06", "#4a2c17", "#7a4a25", "#b06a33", "#d99a4e", "#f0c987", "#c96b2e"],
)

# Bone/sepia gradient used for the fern fronds.
BONE_CMAP = LinearSegmentedColormap.from_list(
    "fossil_bone",
    ["#2e2015", "#5a4632", "#8a7355", "#c9b896", "#eee3ce"],
)

# Cool stone gradient used for the rock-matrix background.
ROCK_CMAP = LinearSegmentedColormap.from_list(
    "fossil_rock",
    ["#1c1712", "#2b241d", "#463a2e", "#6b5a45", "#8f7a5e", "#3a2f24"],
)


# ----------------------------------------------------------------------------
# 2. FRACTAL ROCK TEXTURE (Diamond-Square algorithm)
# ----------------------------------------------------------------------------
def diamond_square(size_power=8, roughness=0.55, seed=None):
    """
    Classic fractal terrain generator. Builds a (2^size_power + 1) square
    grid of values by:
      1. Seeding the four corners randomly.
      2. DIAMOND step: average each square's 4 corners, offset by random
         noise, to set its center.
      3. SQUARE step: average each diamond's 4 neighbors, offset by noise,
         to set the edge midpoints.
      4. Halving the step size and the noise amplitude (`roughness`) each
         pass, so each new pass adds progressively finer, subtler detail
         -- the defining trait of a fractal process (self-similar detail
         repeated at shrinking scales).

    Returns a 2D array normalized to the range [0, 1].
    """
    size = 2 ** size_power + 1
    grid = np.zeros((size, size))
    rng = np.random.default_rng(seed)

    grid[0, 0] = rng.uniform(-1, 1)
    grid[0, -1] = rng.uniform(-1, 1)
    grid[-1, 0] = rng.uniform(-1, 1)
    grid[-1, -1] = rng.uniform(-1, 1)

    step = size - 1
    scale = 1.0

    while step > 1:
        half = step // 2

        # --- diamond step: fill square centers ---
        for y in range(half, size - 1, step):
            for x in range(half, size - 1, step):
                avg = (
                    grid[y - half, x - half] + grid[y - half, x + half] +
                    grid[y + half, x - half] + grid[y + half, x + half]
                ) / 4.0
                grid[y, x] = avg + rng.uniform(-1, 1) * scale

        # --- square step: fill diamond edge midpoints ---
        for y in range(0, size, half):
            x_start = half if (y // half) % 2 == 0 else 0
            for x in range(x_start, size, step):
                total, count = 0.0, 0
                if y - half >= 0:
                    total += grid[y - half, x]; count += 1
                if y + half < size:
                    total += grid[y + half, x]; count += 1
                if x - half >= 0:
                    total += grid[y, x - half]; count += 1
                if x + half < size:
                    total += grid[y, x + half]; count += 1
                grid[y, x] = total / count + rng.uniform(-1, 1) * scale

        step //= 2
        scale *= roughness  # noise shrinks each pass -> finer, subtler detail

    grid = (grid - grid.min()) / (grid.max() - grid.min())
    return grid


# ----------------------------------------------------------------------------
# 3. AMMONITE SPIRAL FRACTAL (logarithmic spiral + chamber "suture" lines)
# ----------------------------------------------------------------------------
def draw_ammonite(ax, center, max_radius, turns=3.0, chambers_per_turn=7,
                   growth_rate=0.15, cmap=AMBER_CMAP, seed=None):
    """
    Draws one fossil ammonite shell as a logarithmic spiral:

        r(theta) = a * e^(b * theta)

    `b` (growth_rate) controls how tightly or loosely the shell coils --
    real ammonites vary between about 0.1 and 0.3. The chambers (visible
    as ribbed "suture" lines on real fossils) are added as short marks
    perpendicular to the shell wall at evenly spaced angles.
    """
    rng = np.random.default_rng(seed)
    theta = np.linspace(0.05, turns * 2 * np.pi, 1500)

    b = growth_rate
    a = max_radius / np.exp(b * theta[-1])
    r = a * np.exp(b * theta)

    x = center[0] + r * np.cos(theta)
    y = center[1] + r * np.sin(theta)

    # Build a color-graded, width-graded line: the shell thickens and
    # brightens as it spirals outward, just as ammonite whorls widen with
    # each new growth stage.
    points = np.array([x, y]).T.reshape(-1, 1, 2)
    segments = np.concatenate([points[:-1], points[1:]], axis=1)
    t_norm = np.linspace(0, 1, len(segments))
    colors = cmap(t_norm)
    linewidths = 1.0 + t_norm * (max_radius * 0.22)

    shell = LineCollection(segments, colors=colors, linewidths=linewidths,
                            capstyle="round", zorder=3)
    ax.add_collection(shell)

    # Faint darker outline along the outer edge for depth/definition.
    outline_r = r * 1.0
    ax.plot(center[0] + outline_r * np.cos(theta),
             center[1] + outline_r * np.sin(theta),
             color="#0d0803", linewidth=0.6, alpha=0.35, zorder=2)

    # Suture / chamber lines: short marks across the shell wall.
    n_chambers = max(4, int(turns * chambers_per_turn))
    chamber_theta = np.linspace(theta[0], theta[-1], n_chambers)
    for t in chamber_theta:
        rr = a * np.exp(b * t)
        px, py = center[0] + rr * np.cos(t), center[1] + rr * np.sin(t)
        # perpendicular ("radial-ish") direction across the shell wall
        nx, ny = np.cos(t), np.sin(t)
        wobble = rng.uniform(0.85, 1.15)
        width = 0.10 * rr * wobble
        ax.plot([px - nx * width, px + nx * width],
                [py - ny * width, py + ny * width],
                color=cmap(min(1.0, t / theta[-1] + 0.05)),
                linewidth=1.0, alpha=0.55, zorder=4)

    # Small dark aperture (opening) at the outer end, like a real shell mouth.
    ax.add_patch(plt.Circle((x[-1], y[-1]), max_radius * 0.055,
                             color="#0d0803", alpha=0.7, zorder=5))


def scatter_ammonites(ax, count, field_size, base_radius, seed=0):
    """
    Places `count` ammonites of shrinking size using the golden angle
    (~137.5 degrees) -- the same spacing rule behind sunflower seed
    heads and pinecones. Because each new shell is smaller than the last
    but follows the same spiral formula, the whole composition is
    SELF-SIMILAR across scales, which is the visual signature of a
    fractal: zoom into any cluster and you find the same shape again,
    just smaller.
    """
    rng = np.random.default_rng(seed)
    golden_angle = np.pi * (3 - np.sqrt(5))

    for i in range(count):
        frac = i / count
        r_pos = (0.18 + 0.82 * np.sqrt(frac)) * field_size * 0.46
        theta_pos = i * golden_angle
        cx = field_size / 2 + r_pos * np.cos(theta_pos)
        cy = field_size / 2 + r_pos * np.sin(theta_pos)

        # Size shrinks with distance from center, echoing natural fossil
        # beds where larger, older specimens anchor the composition.
        size = base_radius * (0.35 + 0.7 * (1 - frac)) * rng.uniform(0.85, 1.1)
        turns = rng.uniform(2.0, 3.0)
        growth = rng.uniform(0.15, 0.20)

        draw_ammonite(ax, (cx, cy), size, turns=turns,
                      growth_rate=growth, seed=seed + i)


# ----------------------------------------------------------------------------
# 4. FOSSIL FERN FROND (recursive binary fractal tree)
# ----------------------------------------------------------------------------
def draw_frond(ax, x, y, angle, length, depth, max_depth, cmap, rng):
    """
    A classic recursive fractal tree, styled to read as a fossilized
    fern frond: each branch spawns two smaller branches at a slightly
    randomized angle and a scaled-down length. Because the same drawing
    rule is applied at every depth, the frond is SELF-SIMILAR -- each
    branch is a smaller copy of the whole.
    """
    if depth == 0 or length < 0.04:
        return

    x2 = x + length * np.cos(angle)
    y2 = y + length * np.sin(angle)

    # Trunk (depth == max_depth) is dark and thick; twig tips (depth -> 0)
    # fade to a pale bone color -- a smaller ratio means an earlier branch.
    color = cmap(1.0 - depth / max_depth)
    linewidth = max(0.35, length * 1.3)
    ax.plot([x, x2], [y, y2], color=color, linewidth=linewidth,
            solid_capstyle="round", alpha=0.85, zorder=2)

    # Two child branches: this 50/50 split with an angular offset is what
    # produces the classic fern/tree silhouette.
    spread = np.radians(rng.uniform(20, 30))
    scale = rng.uniform(0.64, 0.72)

    draw_frond(ax, x2, y2, angle - spread, length * scale, depth - 1,
               max_depth, cmap, rng)
    draw_frond(ax, x2, y2, angle + spread, length * scale, depth - 1,
               max_depth, cmap, rng)


def draw_fossil_frond(ax, root, length, angle, depth, cmap, seed=0):
    rng = np.random.default_rng(seed)
    draw_frond(ax, root[0], root[1], angle, length, depth, depth, cmap, rng)


# ----------------------------------------------------------------------------
# 5. COMPOSE THE FULL ARTWORK
# ----------------------------------------------------------------------------
def make_fossil_fractal_art(save_path="fossil_fractal.png", field_size=10,
                             seed=7, high_res_scale=1):
    plt.style.use("dark_background")
    fig, ax = plt.subplots(figsize=(10, 10), dpi=120 * high_res_scale)
    fig.patch.set_facecolor("#1c1712")
    ax.set_facecolor("#1c1712")

    # --- 1. Fractal rock-texture background ---
    rock = diamond_square(size_power=8, roughness=0.55, seed=seed)
    ax.imshow(rock, extent=(0, field_size, 0, field_size), cmap=ROCK_CMAP,
              origin="lower", zorder=0, alpha=0.9)
    # A soft vignette-like second layer adds subtle depth/mineral banding.
    rock2 = diamond_square(size_power=7, roughness=0.62, seed=seed + 1)
    ax.imshow(rock2, extent=(0, field_size, 0, field_size), cmap=ROCK_CMAP,
              origin="lower", zorder=1, alpha=0.25)

    # --- 2. Fossilized fern fronds tucked into the corners (drawn first,
    #        so the ammonites layer cleanly on top of them) ---
    draw_fossil_frond(ax, (field_size * 0.04, field_size * 0.03),
                      length=field_size * 0.20, angle=np.radians(55),
                      depth=8, cmap=BONE_CMAP, seed=seed + 3)
    draw_fossil_frond(ax, (field_size * 0.97, field_size * 0.02),
                      length=field_size * 0.19, angle=np.radians(128),
                      depth=8, cmap=BONE_CMAP, seed=seed + 4)
    draw_fossil_frond(ax, (field_size * 0.02, field_size * 0.97),
                      length=field_size * 0.17, angle=np.radians(-32),
                      depth=7, cmap=BONE_CMAP, seed=seed + 5)
    draw_fossil_frond(ax, (field_size * 0.98, field_size * 0.96),
                      length=field_size * 0.17, angle=np.radians(-148),
                      depth=7, cmap=BONE_CMAP, seed=seed + 6)

    # --- 3. Scattered ammonite fractals (self-similar at many scales) ---
    scatter_ammonites(ax, count=9, field_size=field_size,
                       base_radius=field_size * 0.11, seed=seed)

    # --- 4. One large focal ammonite, front and center ---
    draw_ammonite(ax, (field_size * 0.5, field_size * 0.52),
                  max_radius=field_size * 0.24, turns=3.0,
                  growth_rate=0.17, seed=seed + 99)

    ax.set_xlim(0, field_size)
    ax.set_ylim(0, field_size)
    ax.set_aspect("equal")
    ax.axis("off")
    fig.suptitle("Fossil Fractals", color="#e8d9b8", fontsize=18,
                 fontweight="bold", y=0.96, family="serif")

    fig.tight_layout()
    fig.savefig(save_path, dpi=fig.dpi, facecolor=fig.get_facecolor())
    print(f"Saved: {save_path}")
    plt.show()


if __name__ == "__main__":
    make_fossil_fractal_art(save_path="fossil_fractal.png", seed=7)
```
